```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
```

```python
np.random.seed(42)
n_samples = 200
X = np.random.uniform(-5, 5, (n_samples, 2))
y = (X[:, 1] > X[:, 0]**2 - 5).astype(int)
print(f"Class distribution: {np.bincount(y)}")
```

```python
plt.figure(figsize=(7, 6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolors='k')
plt.title("Randomly Generated Dataset (Quadratic Distribution)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
clf = SVC(kernel='poly', degree=2, C=1.0)
clf.fit(X, y)
accuracy = clf.score(X, y)
print(f"Training accuracy: {accuracy * 100:.2f}%")
```

```python
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300), np.linspace(y_min, y_max, 300))
Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)
plt.figure(figsize=(7, 6))
plt.contourf(xx, yy, Z, alpha=0.3, cmap='coolwarm')
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolors='k')
plt.title("SVM Decision Boundary (Polynomial Kernel, Degree 2)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
