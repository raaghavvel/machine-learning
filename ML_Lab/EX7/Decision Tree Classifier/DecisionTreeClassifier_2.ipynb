```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt
```

```python
data = {
    'weather': ['sunny', 'sunny', 'sunny', 'sunny', 'sunny',
                'rainy', 'rainy', 'rainy', 'rainy', 'rainy'],
    'activity': ['picnic', 'drinks with colleague', 'errands', 'picnic', 'drinks with colleague',
                 'watch movie', 'watch movie', 'watch movie', 'watch movie', 'watch movie'],
    'outcome': ['go out', 'go out', 'go out', 'go out', 'go out',
                'stay in', 'stay in', 'stay in', 'stay in', 'stay in']
}
df = pd.DataFrame(data)
df
```

```python
le_weather = LabelEncoder()
le_outcome = LabelEncoder()
df['weather_encoded'] = le_weather.fit_transform(df['weather'])   # sunny=1, rainy=0 (order may vary)
df['outcome_encoded'] = le_outcome.fit_transform(df['outcome'])   # go out=0, stay in=1 (order may vary)
df
```

```python
X = df[['weather_encoded']]
y = df['outcome_encoded']
```

```python
clf = DecisionTreeClassifier(criterion='gini', random_state=42)
clf.fit(X, y)
```

```python
plt.figure(figsize=(8, 6))
plot_tree(clf,
          feature_names=['weather'],
          class_names=le_outcome.classes_,
          filled=True,
          rounded=True)
plt.title("Classification Tree: Weather → Go Out / Stay In")
plt.show()
```

```python
predictions = clf.predict(X)
accuracy = (predictions == y).mean()
print(f"Accuracy: {accuracy * 100:.1f}%")
print("Since every rainy day maps to 'stay in' and every sunny day maps to 'go out',")
print("the single split on 'weather' produces 100% pure leaves — no second split needed.")
```
