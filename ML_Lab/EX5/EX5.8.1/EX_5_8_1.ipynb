```python
import pandas as pd
df=pd.read_csv("chicago_taxi_trips.csv")
df=df.dropna()
df=df[(df["trip_miles"]>0) & (df["fare"])]
print(df.shape)
print(df.head())
```

```python
X=df[["trip_miles"]]
Y=df[["fare"]]

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x_test,y_pred)

plt.xlabel("Actual Fare")
plt.ylabel("Predicted Fare")
plt.title("Actual vs Predicted Taxi Fare")
plt.show()
```

```python
plt.scatter(x_test, y_test, color='blue')
plt.plot(x_train, model.predict(x_train), color='red')

plt.title('FARE VS TRIP MILES (Testing Set)')
plt.xlabel('Trip Miles')
plt.ylabel('Fare in Dollars')

plt.show()
```

```python
new_trip = [[5]]

predicted_fare = model.predict(new_trip)

print("Predicted fare:", predicted_fare[0])
```
